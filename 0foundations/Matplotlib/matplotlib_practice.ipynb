```python
import pandas as pd
df_sales = pd.read_excel("linechart.xlsx")
df_sales.head()
```

```python
from matplotlib import pyplot as plt
plt.figure(figsize = (12,4))
plt.plot(df_sales["Quarter"],df_sales["Washing Machine"] , color = "green", label = "Quater")
# plt.plot(df_sales["Quarter"],df_sales["Dishwasher"] , color = "blue",label = "Dishwasher")
# plt.plot(df_sales["Fridge"],df_sales["Washing Machine"] , color = "purple",label = "Washing Machine")
plt.title("potting")
plt.ylabel("Revenue")
plt.xlabel("Financial Quater")


plt.legend()


plt.show()
```

```python
total_sales = df_sales[["Fridge","Dishwasher","Washing Machine"]].sum()
total_sales
```

```python
plt.pie(total_sales,labels = total_sales.index,autopct="%1.1f%%",explode=(0,.5,0),shadow=True,startangle = 170)
plt.show
```
