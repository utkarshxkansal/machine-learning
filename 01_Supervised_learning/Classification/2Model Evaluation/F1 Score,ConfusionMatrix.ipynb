# Classification Metrics: Confusion Matrix and F1 Score

This notebook explores classification evaluation using a confusion matrix and the F1 score. These metrics help assess the types of prediction errors a classifier makes and balance precision and recall.


```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,confusion_matrix
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("car_ownership.csv")
print(df.shape)
df.head(5)
```

```python
# model = LinearRegression()
# model = Ridge(alpha=9)
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

X=df[["monthly_salary"]]
y=df["owns_car"]

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

model.fit(X_train,y_train)

y_pred = model.predict(X_test)

cm = confusion_matrix(y_test, y_pred)

print(cm)



```

```python
model.score(X_test,y_test)
```

```python
import seaborn as sns
classes = ["No Car","Owns car"]
sns.heatmap(cm,annot=True,xticklabels=classes,yticklabels=classes)
plt.xlabel('Predicted Labels')
plt.ylabel('true Labels')
plt.title('confusion Matrix')
plt.show()
```
