# Binary Logistic Regression

This notebook studies binary classification with logistic regression. It covers preparing input features and labels, fitting a classifier, generating predictions, and evaluating the model. Logistic regression estimates the probability of an observation belonging to a class.


```python
import pandas as pd
df = pd.read_csv("car_ownership.csv")
df.head(3)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X = df[['monthly_salary']]
y = df['owns_car']

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
model = LogisticRegression()
model.fit(X_train,y_train)
model.score(X_test,y_test)
```

```python
X_test
```

```python
y_test
```

```python
y_pred = model.predict(X_test)
```

```python
model.predict_proba(X_test)
```

```python
model.coef_,model.intercept_
```

```python
import math 
def sigmoid(x):
    return 1/(1+math.exp(-x))
    
```

```python
sigmoid(9)
```

```python
def prediction_function(salary,model):
    z = model.coef_*salary + model.intercept_
    y = sigmoid(z[0][0])
    return y
    
```
