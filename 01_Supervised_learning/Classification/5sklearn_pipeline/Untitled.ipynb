# Scikit-learn Pipeline with SVM

This notebook builds a scikit-learn `Pipeline` that combines `StandardScaler` and an SVM classifier for the Raisin dataset.

**Workflow:** prepare the data → create a scaling-and-classification pipeline → fit it on the training set → predict and inspect the classification report.

A pipeline keeps preprocessing and model training together, helping make the workflow consistent.


```python
import pandas as pd
df = pd.read_excel("Raisin_Dataset.xlsx")
df.sample(5)
```

```python
df.isna().sum()
```

```python
df.fillna(df.mean(numeric_only=True),inplace =True)
```

```python
df.columns
```

```python
X = df[['Area', 'MajorAxisLength', 'MinorAxisLength', 'Eccentricity',
       'ConvexArea', 'Extent', 'Perimeter']]
y= df['Class']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.svm import SVC
model = SVC(kernel="rbf")
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
from sklearn.metrics import classification_report
report = classification_report(y_test,y_pred)
print(report)
model.n_iter_
```

```python
from sklearn.svm import SVC
model = SVC(kernel="linear")
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
from sklearn.metrics import classification_report
report = classification_report(y_test,y_pred)
print(report)
model.n_iter_
```

```python
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
```

```python
pipeline = Pipeline([
    ('scale',StandardScaler()),
    ('svc',SVC(kernel = "rbf"))
])
```

```python
pipeline.fit(X_train,y_train)
```

```python
y_pred=pipeline.predict(X_test)
from sklearn.metrics import classification_report 
report = classification_report(y_test,y_pred)
model.n_iter_
```
